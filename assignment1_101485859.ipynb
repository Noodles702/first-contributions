```python
"""

Author: Anthony Truong
Assignment: #1

"""

# Variable Declarations
gym_member = "Alex Alliton"  # String
preferred_weight_kg = 20.5  # Float
highest_reps = 25  # Integer
membership_active = True  # Boolean

# Dictionary: Stores friends' workout times in the order: "name": (yoga, running, weightlifting)
workout_stats = {
    "Alex": (30, 45, 20),
    "Obama": (60, 20, 60),
    "Stark": (40, 20, 60)
}

# Calculate total workout times in a separate dictionary to avoid modification errors
total_workouts = {f"{friend}_Total": sum(minutes) for friend, minutes in workout_stats.items()}
workout_stats.update(total_workouts)


workout_list = [list(minutes) for friend, minutes in workout_stats.items() if "Total" not in friend]


print("\nYoga and Running times for all friends:")
for friend, minutes in workout_stats.items():
    if "Total" not in friend:
        print(f"{friend}: {minutes[:2]}")


print("\nWeightlifting times for last two friends:")
friends_list = [friend for friend in workout_stats if "Total" not in friend]
for friend in friends_list[-2:]:
    print(f"{friend}: {workout_stats[friend][2]}")


for friend, total in total_workouts.items():
    if total >= 120:
        print(f"\nGreat job staying active, {friend.replace('_Total', '')}!")


friend_name = input("\nEnter a friend's name: ").strip()
if friend_name in workout_stats and isinstance(workout_stats[friend_name], tuple):
    minutes = workout_stats[friend_name]
    total_minutes = total_workouts.get(f"{friend_name}_Total", 0)
    print(f"\n{friend_name}'s workout time:")
    print(f"Yoga: {minutes[0]}, Running: {minutes[1]}, Weightlifting: {minutes[2]}")
    print(f"Total workout time: {total_minutes}")
else:
    print(f"\nFriend {friend_name} not found in the records.")


highest_friend = max(total_workouts, key=total_workouts.get).replace("_Total", "")
lowest_friend = min(total_workouts, key=total_workouts.get).replace("_Total", "")

print(f"\n Person with the highest total workout time: {highest_friend}")
print(f" Person with the lowest total workout time: {lowest_friend}")
```
